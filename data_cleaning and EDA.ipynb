```python
df['Booking_Date']=pd.to_datetime(df['Booking_Date'],format='mixed',dayfirst=True)
```

```python
df['Show_Date']=pd.to_datetime(df['Show_Date'],format='mixed',dayfirst=True)
```

```python
columns=[ 'City', 'Theatre', 'Movie',
       'Genre', 'Language', 'Screen_Format', 'Show_Time', 'Seats_Booked',
       'Ticket_Price', 'Gross_Ticket_Amount', 'Discount_Pct',
       'Discount_Amount', 'Convenience_Fee', 'Payment_Method',
       'Booking_Status']
```

```python
for col in columns:
    df[col] = df[col].astype(str).str.strip().str.lower()

    print(f"\nColumn: {col}")
    print(df[col].unique())
```

```python
numeric_columns = [
    'Seats_Booked',
    'Ticket_Price',
    'Gross_Ticket_Amount',
    'Discount_Pct',
    'Discount_Amount',
    'Convenience_Fee'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
text_columns = [
    'Booking_ID',
    'City',
    'Theatre',
    'Movie',
    'Genre',
    'Language',
    'Screen_Format',
    'Show_Time',
    'Payment_Method',
    'Booking_Status'
]

for col in text_columns:
    df[col] = df[col].str.strip().str.lower()
```

```python
df.info()
```

```python
print("Total duplicates:", df.duplicated().sum())

df = df.drop_duplicates(subset=['Booking_ID'])
```

```python
plt.figure(figsize=(12,5))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values Analysis")
plt.show()
```

```python
movie_bookings = df.groupby('Movie')['Booking_ID'].count().sort_values(ascending=False)

print(movie_bookings)
```

```python
plt.figure(figsize=(12,6))

movie_bookings.plot(kind='bar')

plt.title("Movie-wise Total Bookings")
plt.xlabel("Movies")
plt.ylabel("Total Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python
city_bookings = df.groupby('City')['Booking_ID'].count().sort_values(ascending=False)

print(city_bookings)
```

```python
plt.figure(figsize=(12,6))

city_bookings.plot(kind='bar')

plt.title("City-wise Booking Analysis")
plt.xlabel("City")
plt.ylabel("Total Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python
genre_bookings = df.groupby('Genre')['Booking_ID'].count().sort_values(ascending=False)

print(genre_bookings)
```

```python
plt.figure(figsize=(8,8))

genre_bookings.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Movie Genre Distribution")
plt.ylabel("")
plt.show()
```

```python
df['Net_Revenue'] = (
    df['Gross_Ticket_Amount'] - df['Discount_Amount']
)
```

```python
confirmed_df = df[df['Booking_Status'] == 'confirmed'].copy()
```

```python
city_revenue = confirmed_df.groupby('City')['Net_Revenue'].sum().sort_values(ascending=False)

print(city_revenue)
```

```python
plt.figure(figsize=(12,6))

city_revenue.plot(kind='bar')

plt.title("City-wise Revenue Analysis")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
payment_analysis = df['Payment_Method'].value_counts()

print(payment_analysis)
```

```python
plt.figure(figsize=(8,8))

payment_analysis.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Payment Method Distribution")
plt.ylabel("")
plt.show()
```

```python
showtime_analysis = df.groupby('Show_Time')['Booking_ID'].count()

print(showtime_analysis)
```

```python
plt.figure(figsize=(10,5))

showtime_analysis.plot(kind='bar')

plt.title("Bookings by Show Time")
plt.xlabel("Show Time")
plt.ylabel("Total Bookings")
plt.xticks(rotation=0)
plt.show()
```

```python
theatre_analysis = confirmed_df.groupby('Theatre')['Net_Revenue'].sum().sort_values(ascending=False)

print(theatre_analysis)
```

```python
plt.figure(figsize=(12,6))

theatre_analysis.plot(kind='bar')

plt.title("Theatre-wise Revenue Analysis")
plt.xlabel("Theatre")
plt.ylabel("Revenue")
plt.xticks(rotation=75)
plt.show()
```

```python
df['Booking_Month'] = df['Booking_Date'].dt.to_period('M')

monthly_bookings = df.groupby('Booking_Month')['Booking_ID'].count()

print(monthly_bookings)
```

```python
plt.figure(figsize=(12,6))

monthly_bookings.plot(kind='line', marker='o')

plt.title("Monthly Booking Trends")
plt.xlabel("Month")
plt.ylabel("Total Bookings")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(10,7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Create final revenue column
df['Net_Revenue'] = (
    df['Gross_Ticket_Amount'] - df['Discount_Amount']
)

# Export cleaned dataset
df.to_csv(
    "BookMyShow_Cleaned_Data.csv",
    index=False
)

print("Dataset exported successfully!")
```
